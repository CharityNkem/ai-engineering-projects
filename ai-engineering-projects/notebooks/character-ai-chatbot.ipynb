# Character AI Chatbot

A conversational chatbot that takes on a **defined character/persona** and holds a coherent, in-character conversation using the OpenAI API.

**What it demonstrates**
- Designing a system prompt to shape an assistant's personality and voice
- Managing multi-turn conversation state (message history)
- Working directly with the OpenAI chat completions API

**Stack:** Python · OpenAI API

```python
import os
from dotenv import load_dotenv
load_dotenv()
openai_api_key = os.getenv("OPENAI_API_KEY")

openai = OpenAI(api_key = openai_api_key)
print("OpenAI key connected successfully!")
```

```python
my_message = "Write a message to my husband who is a medical doctor and explain AI Engineering to him in a simple way that he can understand."
response = openai.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": my_message}])
ai_message = response.choices[0].message.content
print(ai_message)
```
